####**Questão 3**  

***Fermentação alcoólica com cinética dependente da temperatura*** - *Um processo em
batelada converte glicose em etanol e CO2. A taxa de consumo da glicose depende da temperatura segundo a expressão:*  

$$
k(T) = 464{,}8 \cdot \exp\left( \frac{-3985}{T(^{\circ}\mathrm{C}) + 273} \right) \ [\text{L/min}]
$$  

*A cinética é de primeira ordem, tal que:*  

$$
\frac{dS}{dt} = -k(T) \cdot S
$$  

*Considerando concentração inicial S(t = 0) de glicose igual a 100 g/L e temperatura constante de 35°C, implemente uma solução computacional para simular a concentração de glicose ao longo de 24 horas. Qual o valor final da concentração de glicose? O que acontece se a temperatura for aumentada para 50°C?*

```python
"""
====
Back
====
"""
import numpy as np
from scipy.integrate import solve_ivp
from dataclasses import dataclass

"""
Condições iniciais
"""
@dataclass
class Parametros:
  S0: float = 100  # Concentração inicial (g/L)
  T: float = 35  # Temperatura constante (°C)
  t0: float = 0  # Tempo inicial (min)
  tf: float = 24*60  # Tempo final (min)

"""
Equações
"""
def k(params):
  Temperatura = params.T
  return 464.8*np.exp(-3985/(Temperatura + 273))
# Função para k(T)

def dSdt(t, S, T):
    return -k(params)*S
# Equação diferencial: dS/dt = -k(T)*S

def simula_fermentacao(params):
  S0 = params.S0
  tempo = [params.t0, params.tf]
  teval = np.linspace(tempo[0], tempo[-1], 100)
  sol = solve_ivp(dSdt, tempo, [S0], t_eval = teval,args = (params,), method="BDF")
  return sol.t, sol.y[0]
# Resolve a equação diferencial e simula a fermentação
```

```python
"""
====
Front
====
"""
conjunto = [Parametros(),Parametros(T=50)]
# O Parametros() simula o processo com T = 35°C, assim como dado na condição inicial, o Parametros(T=50) muda a temperatura de 35°C para 50°C

for params in conjunto:
  t, S = simula_fermentacao(params)
  print(f"A concentração final de glicose depois de 24h para {params.T}°C é {S[-1]:.3f} g/L")
```

####**Questão 4**  

***Equilíbrio em coluna de destilação com seletividade dependente da temperatura*** *- Um processo modelado com três estágios separa uma mistura etanol/água. A fração molar líquida de etanol (xi) e a temperatura (Ti) em cada estágio são conhecidas. O equilíbrio líquido-vapor é dado por:*  

$$
y_i = \frac{\alpha(T_i) \cdot x_i}{1 + \left[\alpha(T_i) - 1\right] \cdot x_i}
$$  

*com a seguinte expressão linear para a seletividade α(T) = 2.5 − 0.01 · (T − 78).*  

*Os dados para os três estágios são:*  

| Estágio        | \(x_i\) | \(T_i\) (°C) |
|----------------|--------:|-------------:|
| 1 (topo)       |   0.20  |        78    |
| 2 (alimentação)|   0.40  |        80    |
| 3 (fundo)      |   0.60  |        85    |  

*Escreva um código modular que calcule e exiba a seletividade αi em cada estágio e a fração yi com base nos dados acima.*

```python
"""
====
Back
====
"""
import numpy as np
from scipy.integrate import solve_ivp
from dataclasses import dataclass

"""
Condições iniciais
"""
@dataclass
class Parametros:
  nome: str
  x_i: float
  T_i: float

"""
Equações
"""
def seletividade(params):
  T = params.T_i
  return 2.5 - 0.01 * (T - 78)
# Calcula a seletividade α(T)

def equilibrio(params):
  a = seletividade(params)
  xi = params.x_i
  return (a * xi) / (1 + (a - 1) * xi)
#Calcula o equilíbrio líquido-vapor
```

```python
"""
====
Front
====
"""
conjunto = [Parametros(nome='topo',x_i=0.2, T_i=78 ),
            Parametros(nome='alimentação', x_i=0.4, T_i=80),
            Parametros(nome='fundo', x_i=0.6, T_i=85)]
# O primeiro Parametros(...) simula o processo com as condições do topo, o segundo simula com as condições da alimentação e o terceiro simula com as condições do fundo.

for params in conjunto:
  y_i = equilibrio(params)
  print(f"O equilíbrio líquido-vapor para {params.nome} é {y_i:.3f}")
```
